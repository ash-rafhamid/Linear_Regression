```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
df = load_diabetes()
```

```python
df = pd.DataFrame(data=df.data, columns=df.feature_names);
```

```python
df['target'] = load_diabetes().target
```

```python
X = df[['bmi']].values
y = df['target'].values.reshape(-1 ,1)
X_train,X_test ,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=2);
```

```python
class LR:
    def __init__(self, epoch, learning_rate):
        self.epoch = epoch
        self.learning_rate = learning_rate
        self.coef = 0
        self.intercept = 0

    def fit(self, X_train, y_train):
        X_train = X_train.flatten()
        y_train = y_train.flatten()
        m = len(X_train)

        coef = 0
        intercept = 0

        for i in range(self.epoch):
            y_pred = coef * X_train + intercept
            error = y_pred - y_train

            gradient_m = (2/m) * np.dot(error, X_train)
            gradient_b = (2/m) * np.sum(error)

            coef -= self.learning_rate * gradient_m
            intercept -= self.learning_rate * gradient_b

        self.coef = coef
        self.intercept = intercept

    def predict(self, X):
        return self.coef * X + self.intercept
```

```python
model = LR(epoch=1000, learning_rate=0.989999)
model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Intercept:", model.intercept)
print("Coefficient:", model.coef)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
```
